```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Define the three points (x, y, z)
points = np.array([
    [1, 1, 2],  # Point 1: (x=1, y=1, z=2)
    [2, 2, 3],  # Point 2: (x=2, y=2, z=3)
    [3, 3, 5]   # Point 3: (x=3, y=3, z=5)
])

# Extract x, y, and z
X = points[:, :2]  # Input features: x and y
y = points[:, 2]   # Output target: z

# Fit a model (e.g., Linear Regression)
model = LinearRegression()
model.fit(X, y)

# Test interpolation for new points
new_points = np.array([
    [1.5, 1.5],  # Between Point 1 and Point 2
    [2.5, 2.5],  # Between Point 2 and Point 3
    [3, 3]       # Revisit Point 3
])

# Predict z values for new points
predicted_z = model.predict(new_points)

# Print results
for i, (inp, pred) in enumerate(zip(new_points, predicted_z)):
    print(f"Interpolated Point {i+1}: Input (x={inp[0]}, y={inp[1]}) -> Predicted z={pred:.2f}")

# Visualization (Optional)
plt.scatter(points[:, 0], points[:, 2], color='blue', label='Original Points (x, z)')
plt.scatter(new_points[:, 0], predicted_z, color='red', label='Interpolated Points')
plt.plot(points[:, 0], y, color='green', linestyle='--', label='Fitted Line')
plt.xlabel('x')
plt.ylabel('z')
plt.legend()
plt.title('Three-Point Interpolation')
plt.show()
```
